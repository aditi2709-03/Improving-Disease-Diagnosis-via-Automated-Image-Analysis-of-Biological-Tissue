```python
# output directory

video_filenames = ["Training video 2.mp4", "Training video 3.mp4"]
output_base_dir = "/content/extracted_frames"
os.makedirs(output_base_dir, exist_ok=True)
```

```python
FRAME_RATE = 30
```

```python
def extract_frames(video_path, output_dir, frame_rate=30):
    cap = cv2.VideoCapture(video_path)
    if not cap.isOpened():
        print(f"Failed to open {video_path}")
        return

    fps = cap.get(cv2.CAP_PROP_FPS)
    interval = int(fps // frame_rate)
    frame_count = 0
    saved_count = 0

    os.makedirs(output_dir, exist_ok=True)

    while True:
        ret, frame = cap.read()
        if not ret:
            break

        if frame_count % interval == 0:
            frame_name = f"{output_dir}/frame_{saved_count:05d}.jpg"
            cv2.imwrite(frame_name, frame)
            saved_count += 1

        frame_count += 1

    cap.release()
    print(f"Extracted {saved_count} frames from {video_path} into {output_dir}")
```

```python
for video_file in video_filenames:
    base_name = os.path.splitext(os.path.basename(video_file))[0]
    output_folder = os.path.join(output_base_dir, base_name)
    extract_frames(video_file, output_folder, frame_rate=FRAME_RATE)
```

```python
import shutil

video_names = ["Training video 2", "Training video 3"]
base_path = "/content/extracted_frames"

for name in video_names:
    folder_path = f"{base_path}/{name}"
    zip_path = f"/content/{name.replace(' ', '_')}.zip"
    shutil.make_archive(zip_path.replace('.zip', ''), 'zip', folder_path)
    print(f"Zipped: {zip_path}")
```
